## 3. Summary Functions and Maps

**3.1** Use `describe()` no DataFrame `vendas` e observe as estatísticas.

**3.2** Calcule a média (`mean`) e a mediana (`median`) da coluna `preco`.

**3.3** Descubra quantas categorias únicas existem na coluna `categoria` (use `unique()` e `value_counts()`).

**3.4** Use `map()` para criar uma coluna `preco_com_desconto` aplicando 10% de desconto sobre `preco`.

**3.5** Use `apply()` com uma função própria para criar uma coluna `faixa`, classificando cada produto como `"barato"` (preco < 200), `"médio"` (200 <= preco < 1000) ou `"caro"` (preco >= 1000).

```python
# 3.1
vendas_descricao = vendas.describe()
vendas_descricao
```

```python
# 3.2
vendas_media_mediana = vendas["preco"].agg(["mean", "median"])
vendas_media_mediana
```

```python
# 3.3
vendas_unicas = vendas["categoria"].unique().value_counts()
vendas_unicas
```

```python
# 3.4
preco_com_desconto = vendas.assign(preco_desconto=vendas["preco"].map(lambda p: p * 0.9))
preco_com_desconto
```

```python
# 3.5
def classificar_faixa(preco):
    if preco < 200:
        return "barato"
    elif preco < 1000:
        return "médio"
    else:
        return "caro"

vendas["faixa"] = vendas["preco"].apply(classificar_faixa)
vendas
```

```python
# 6.3
vendas_p2 = pd.DataFrame({
    "produto": ["Caixa de Som", "Pendrive", "HD Externo"],
    "categoria": ["Eletrônicos", "Eletrônicos", "Eletrônicos"],
    "preco": [150, 800, 1200],
    "quantidade": [15, 10, 5]
})
vendas_completo = pd.concat([vendas[["produto", "categoria", "preco", "quantidade"]], vendas_p2], ignore_index=True)
vendas_completo

```
