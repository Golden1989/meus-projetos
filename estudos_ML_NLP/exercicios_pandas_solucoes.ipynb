```python
import pandas as pd
import numpy as np

vendas = pd.DataFrame({
    "produto": ["Notebook", "Mouse", "Teclado", "Monitor", "Cadeira",
                "Mesa", "Headset", "Webcam", "Impressora", "Roteador"],
    "categoria": ["Eletrônicos", "Eletrônicos", "Eletrônicos", "Eletrônicos", "Móveis",
                  "Móveis", "Eletrônicos", "Eletrônicos", "Eletrônicos", "Eletrônicos"],
    "preco": [3500, 80, 150, 1200, 450, 600, 200, 120, 900, 250],
    "quantidade": [10, 50, 30, 15, 8, 5, 25, 40, 12, 20]
})
vendas
```

## 3. Summary Functions and Maps

```python
# 3.1
vendas.describe()
```

```python
# 3.2
vendas["preco"].mean(), vendas["preco"].median()
```

```python
# 3.3
vendas["categoria"].unique(), vendas["categoria"].value_counts()
```

```python
# 3.4
vendas["preco_com_desconto"] = vendas["preco"].map(lambda p: p * 0.9)
vendas
```

```python
# 3.5
def classificar_faixa(preco):
    if preco < 200:
        return "barato"
    elif preco < 1000:
        return "médio"
    else:
        return "caro"

vendas["faixa"] = vendas["preco"].apply(classificar_faixa)
vendas
```

## 4. Grouping and Sorting

```python
# 4.1
vendas.groupby("categoria")["quantidade"].sum()
```

```python
# 4.2
vendas.groupby("categoria").agg({"preco": "mean", "produto": "count"})
```

```python
# 4.3
vendas.sort_values("preco", ascending=False)
```

```python
# 4.4
vendas.groupby("categoria")["quantidade"].sum().sort_index()
```

## 5. Data Types and Missing Values

```python
# 5.1
vendas_com_nulos = vendas.copy()
vendas_com_nulos.loc[[2, 5], "preco"] = np.nan
vendas_com_nulos
```

```python
# 5.2
vendas_com_nulos.isnull().sum()
```

```python
# 5.3
media_preco = vendas_com_nulos["preco"].mean()
vendas_com_nulos["preco"] = vendas_com_nulos["preco"].fillna(media_preco)
vendas_com_nulos
```

```python
# 5.4
print(vendas.dtypes)
vendas["quantidade"] = vendas["quantidade"].astype(float)
vendas.dtypes
```

## 6. Renaming and Combining

```python
# 6.1
vendas_renomeado = vendas.rename(columns={"preco": "preco_unitario"})
vendas_renomeado
```

```python
# 6.2
vendas_renomeado = vendas.rename_axis("id_venda")
vendas_renomeado
```

```python
# 6.3
vendas_2 = pd.DataFrame({
    "produto": ["Caixa de Som", "Pendrive", "HD Externo"],
    "categoria": ["Eletrônicos", "Eletrônicos", "Eletrônicos"],
    "preco": [300, 40, 350],
    "quantidade": [18, 60, 22]
})
vendas_completo = pd.concat([vendas[["produto", "categoria", "preco", "quantidade"]], vendas_2], ignore_index=True)
vendas_completo
```
