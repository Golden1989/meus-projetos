## 3.2 Definir o modelo

- `self.rnn = nn.RNN(input_size=embed_dim, hidden_size=32, batch_first=True)`
  (use `config["embed_dim"]`)
- `self.fc = nn.Linear(32, 1)`

No `forward(self, x)`:
1. `emb = self.embedding(x)`
2. `out, hidden = self.rnn(emb)`
3. `logits = self.fc(hidden.squeeze(0))`
4. retorne `torch.sigmoid(logits).squeeze(1)`
