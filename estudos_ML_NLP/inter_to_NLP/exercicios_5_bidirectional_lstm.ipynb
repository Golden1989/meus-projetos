## 5.2 Definir o modelo Bidirectional LSTM

Crie `SMSBiLSTM(nn.Module)`:
- `self.lstm = nn.LSTM(input_size=embed_dim, hidden_size=32, batch_first=True, bidirectional=True)`
- `self.fc = nn.Linear(32 * 2, 1)` (o dobro do `hidden_size`, pois temos
  uma leitura "pra frente" e uma "pra trás")

pra frente, índice 1 = leitura pra trás). Concatene as duas com
`torch.cat([hidden[0], hidden[1]], dim=1)` antes de passar pela camada
