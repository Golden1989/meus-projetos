```python
# 1.1
print(sms.shape)
print(sms["label"].value_counts())
print(sms["label"].value_counts(normalize=True))
```

```python
# 1.2
sms["text_lower"] = sms["text"].str.lower()
sms[["text", "text_lower"]].head()
```

```python
# 1.3
def remove_punct_numbers(texto):
    return re.sub(r"[^a-z\s]", "", texto)

sms["text_no_punct"] = sms["text_lower"].apply(remove_punct_numbers)
sms[["text_lower", "text_no_punct"]].head()
```

```python
# 1.4
sms["tokens"] = sms["text_no_punct"].str.split()
sms["tokens"].head()
```

```python
# 1.5
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

sms["tokens_no_stopwords"] = sms["tokens"].apply(
    lambda palavras: [p for p in palavras if p not in ENGLISH_STOP_WORDS]
)
sms["tokens_no_stopwords"].head()
```

```python
# 1.6
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
sms["tokens_stemmed"] = sms["tokens_no_stopwords"].apply(
    lambda palavras: [stemmer.stem(p) for p in palavras]
)
sms["tokens_stemmed"].head()
```

```python
# 1.7
sms["clean_text"] = sms["tokens_stemmed"].apply(lambda palavras: " ".join(palavras))
sms[["text", "clean_text"]].head()
```

```python
# 1.8
sms[["label", "text", "clean_text"]].to_csv("sms_clean.csv", index=False)
```
