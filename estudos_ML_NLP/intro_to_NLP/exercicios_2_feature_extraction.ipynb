# Módulo 2 — Feature Extraction

Continuação do Módulo 1. Aqui carregamos o `sms_clean.csv` gerado lá, transformamos o texto em números (Bag-of-Words e TF-IDF) e separamos treino/teste. No final, salvamos `train.csv` e `test.csv`, que o Módulo 3 (Modeling with XGBoost) vai usar.

Se você ainda não rodou o Módulo 1, rode ele primeiro para gerar o `sms_clean.csv`.

Tente resolver antes de olhar `exercicios_2_feature_extraction_solucoes.ipynb`.

```python
import pandas as pd

sms = pd.read_csv("sms_clean.csv")
sms.head()
```

## 2.1 Checar os dados

Verifique se há valores nulos em `sms` (`isnull().sum()`). Se houver (pode acontecer se algum texto ficou vazio depois da limpeza), remova essas linhas com `dropna()`, guardando o resultado em `sms`.

```python
# 2.1
print(sms.isnull().sum())
sms = sms.dropna()
```

## 2.2 Preparar X e y

Crie `X_text = sms["clean_text"]`.

Crie `y`, convertendo a coluna `label` para números: `"ham"` vira `0` e `"spam"` vira `1` (use `.map({"ham": 0, "spam": 1})`).

```python
# 2.2
X_text = sms["clean_text"]
y = sms["label"].map({"ham": 0, "spam": 1})
```

## 2.3 Dividir treino e teste

Use `train_test_split` (de `sklearn.model_selection`) para criar `X_train_text`, `X_test_text`, `y_train`, `y_test`.

Use `test_size=0.25`, `random_state=1` e `stratify=y` (o `stratify` garante que a proporção de spam/ham fique igual nos dois conjuntos — importante numa base desbalanceada como essa).

```python
# 2.3
from sklearn.model_selection import train_test_split

X_train_text, X_test_text, y_train, y_test = train_test_split(X_text, y, test_size=0.25, random_state=1, stratify=y)
print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```

## 2.4 Bag-of-Words

Use `CountVectorizer` (de `sklearn.feature_extraction.text`). Crie o vetorizador, treine com `fit_transform` em `X_train_text` (gerando `X_train_bow`), e use `transform` (não `fit_transform`!) em `X_test_text` (gerando `X_test_bow`).

Imprima quantas palavras únicas entraram no vocabulário (`len(vectorizer.get_feature_names_out())`).

```python
# 2.4
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()
X_train_bow = count_vectorizer.fit_transform(X_train_text)
X_test_bow = count_vectorizer.transform(X_test_text)
print("Tamanho do vocabulário:", len(count_vectorizer.get_feature_names_out()))
print(X_test_bow.toarray())
```

## 2.5 TF-IDF

Repita o processo do 2.4, agora usando `TfidfVectorizer` no lugar de `CountVectorizer`, gerando `X_train_tfidf` e `X_test_tfidf`.

```python
# 2.5
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train_text)
X_test_tfidf = tfidf_vectorizer.transform(X_test_text)
print("Tamanho do vocabulário:", len(tfidf_vectorizer.get_feature_names_out()))
```

## 2.6 Explorar o vocabulário

Some as colunas de `X_train_bow` (some quantas vezes cada palavra aparece no total) e descubra as 10 palavras mais frequentes.

Dica: `X_train_bow.sum(axis=0)` dá a soma por coluna; combine com `vectorizer.get_feature_names_out()` numa Series do pandas pra ordenar com `sort_values()`.

```python
# 2.6
word_count = pd.Series( X_train_bow.sum(axis = 0).A1, index = count_vectorizer.get_feature_names_out()) 
word_count.sort_values(ascending = False).head(10)
```

## 2.7 Salvar para o próximo módulo

O Módulo 3 vai vetorizar o texto de novo (é uma boa prática treinar o vetorizador só com dados de treino, no próprio módulo que faz o treino do modelo). Por isso aqui a gente salva só o **texto limpo e o rótulo**, não os números já vetorizados.

Crie um DataFrame `train_df` com colunas `clean_text` (de `X_train_text`) e `label` (de `y_train`), e salve em `train.csv` (`index=False`).

Faça o mesmo para `test_df` com `X_test_text`/`y_test`, salvando em `test.csv`.

```python
# 2.7
train_df = pd.DataFrame({"clean_text": X_train_text, "label": y_train})
test_df = pd.DataFrame({"clean_text": X_test_text, "label": y_test})

train_df.to_csv("train.csv", index=False)
test_df.to_csv("test.csv", index=False)
```
