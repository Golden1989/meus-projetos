# Gabarito — Módulo 2: Feature Extraction

```python
import pandas as pd

sms = pd.read_csv("sms_clean.csv")
sms.head()
```

```python
# 2.1
print(sms.isnull().sum())
sms = sms.dropna()
```

```python
# 2.2
X_text = sms["clean_text"]
y = sms["label"].map({"ham": 0, "spam": 1})
```

```python
# 2.3
from sklearn.model_selection import train_test_split

X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text, y, test_size=0.25, random_state=1, stratify=y
)
print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```

```python
# 2.4
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()
X_train_bow = count_vectorizer.fit_transform(X_train_text)
X_test_bow = count_vectorizer.transform(X_test_text)
print("Tamanho do vocabulário:", len(count_vectorizer.get_feature_names_out()))
```

```python
# 2.5
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train_text)
X_test_tfidf = tfidf_vectorizer.transform(X_test_text)
```

```python
# 2.6
word_counts = pd.Series(
    X_train_bow.sum(axis=0).A1,
    index=count_vectorizer.get_feature_names_out()
)
word_counts.sort_values(ascending=False).head(10)
```

```python
# 2.7
train_df = pd.DataFrame({"clean_text": X_train_text, "label": y_train})
test_df = pd.DataFrame({"clean_text": X_test_text, "label": y_test})

train_df.to_csv("train.csv", index=False)
test_df.to_csv("test.csv", index=False)
```
