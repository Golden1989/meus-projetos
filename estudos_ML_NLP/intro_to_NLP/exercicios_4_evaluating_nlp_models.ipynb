```python
import pandas as pd

results = pd.read_csv("predictions.csv")
results.head()
```

## 4.1 Acurácia

Use `accuracy_score` (de `sklearn.metrics`) para calcular a acurácia entre `results["y_true"]` e `results["y_pred"]`.

```python
# 4.1
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(results["y_true"], results["y_pred"])
accuracy
```

## 4.2 Precision, Recall e F1

Use `precision_score`, `recall_score` e `f1_score` (de `sklearn.metrics`) para calcular essas 3 métricas, considerando a classe "spam" (`1`) como a classe positiva (`pos_label=1`).

```python
# 4.2
from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(results["y_true"], results["y_pred"], pos_label=1)
recall = recall_score(results["y_true"], results["y_pred"], pos_label=1)
f1 = f1_score(results["y_true"], results["y_pred"], pos_label=1)

print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1-score: {f1:.2f}")
```

## 4.3 Classification report

Use `classification_report` (de `sklearn.metrics`) para imprimir um resumo com todas as métricas por classe de uma vez.

```python
# 4.3
from sklearn.metrics import classification_report
print(classification_report(results["y_true"], results["y_pred"], target_names=["ham", "spam"]))
```

## 4.4 Matriz de confusão

Use `confusion_matrix` (de `sklearn.metrics`) para gerar a matriz de confusão entre `y_true` e `y_pred`.

Depois, em uma célula de markdown (ou comentário), escreva com suas palavras: quantos falsos positivos (ham classificado como spam) e falsos negativos (spam classificado como ham) o modelo teve?

```python
# 4.4
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(results["y_true"], results["y_pred"])
cm

```

## 4.5 ROC-AUC

Use `roc_auc_score` (de `sklearn.metrics`) para calcular o ROC-AUC, usando `y_true` e `y_proba` (a probabilidade prevista, não a classe).

```python
# 4.5
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(results["y_true"], results["y_proba"])
roc_auc

```
