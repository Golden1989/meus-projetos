# Gabarito — Módulo 4: Evaluating NLP Models

````python
import pandas as pd

results = pd.read_csv("predictions.csv")
results.head()
````

````python
# 4.1
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(results["y_true"], results["y_pred"])
accuracy
````

````python
# 4.2
from sklearn.metrics import precision_score, recall_score, f1_score

precision = precision_score(results["y_true"], results["y_pred"], pos_label=1)
recall = recall_score(results["y_true"], results["y_pred"], pos_label=1)
f1 = f1_score(results["y_true"], results["y_pred"], pos_label=1)

print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1-score: {f1:.2f}")
````

````python
# 4.3
from sklearn.metrics import classification_report

print(classification_report(results["y_true"], results["y_pred"], target_names=["ham", "spam"]))
````

````python
# 4.4
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(results["y_true"], results["y_pred"])
cm
````

A matriz de confusão tem o formato:

```
[[TN, FP],
 [FN, TP]]
```

- `FP` (linha 0, coluna 1): mensagens "ham" classificadas como "spam" (falso positivo)
- `FN` (linha 1, coluna 0): mensagens "spam" classificadas como "ham" (falso negativo — o mais problemático, já que deixa spam passar)

````python
# 4.5
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(results["y_true"], results["y_proba"])
roc_auc
````

## 4.6 Reflexão (exemplo de resposta)

Como a base tem muito mais "ham" do que "spam", um modelo que sempre prevê "ham" já acertaria a maioria das vezes e teria uma acurácia alta, mesmo sem aprender nada útil sobre o que é spam. Por isso, acurácia sozinha esconde esse problema.

Métricas mais úteis aqui são **recall** da classe spam (quantos spams reais o modelo conseguiu pegar) e **precision** da classe spam (de tudo que o modelo marcou como spam, quanto era spam de verdade), além do **F1-score**, que equilibra as duas. O **ROC-AUC** também ajuda porque avalia a capacidade do modelo de separar as classes independente do limiar de decisão escolhido.
