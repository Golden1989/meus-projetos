```python
import pandas as pd
import numpy as np

np.random.seed(1)
n = 60

casas = pd.DataFrame({
    "area_m2": np.random.randint(40, 250, n),
    "quartos": np.random.randint(1, 5, n),
    "banheiros": np.random.randint(1, 4, n),
    "idade_imovel": np.random.randint(0, 40, n),
    "distancia_centro_km": np.round(np.random.uniform(0.5, 20, n), 1),
})

casas["preco"] = (
    casas["area_m2"] * 3500
    + casas["quartos"] * 15000
    + casas["banheiros"] * 8000
    - casas["idade_imovel"] * 1000
    - casas["distancia_centro_km"] * 2000
    + np.random.normal(0, 10000, n)
).round(2)

# valores faltando de propósito, pra praticar tratamento de dados
casas.loc[[3, 15, 30, 45], "idade_imovel"] = np.nan

casas.head()
```

## 4. Underfitting and Overfitting

**4.1** Escreva uma função `get_mae(max_leaf_nodes, X_train, X_val, y_train, y_val)` que cria um `DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=1)`, treina, prevê em `X_val` e retorna o `mean_absolute_error`.

**4.2** Teste a função para os valores `[5, 25, 50, 100, 250, 500]` de `max_leaf_nodes` e descubra qual apresenta o menor MAE.

**4.3** Treine um modelo final usando o melhor `max_leaf_nodes` encontrado.

```python
# 4.1
def get_mae(max_leaf_nodes, X_train, X_val, y_train, y_val):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=1)
    model.fit(X_train, y_train)
    preds_val = model.predict(X_val)
    mae = mean_absolute_error(y_val, preds_val)
    return mae

```

```python
# 4.2
candidates = [5, 25, 50, 100, 250, 500]
mae = {n : get_mae(n, X_train, X_val, y_train, y_val) for n in candidates}
print(mae)


best_max_leaf_nodes = min(mae, key=mae.get)
best_max_leaf_nodes
```

```python
# 4.3
modelo_final = DecisionTreeRegressor(max_leaf_nodes=best_max_leaf_nodes, random_state=1)
modelo_final.fit(X, y)
```

## 5. Random Forests

**5.1** Importe `RandomForestRegressor` de `sklearn.ensemble`, crie o modelo com `random_state=1`, treine com `X_train`/`y_train` e calcule o MAE nas previsões de `X_val`.

**5.2** Compare o MAE da Random Forest com o melhor MAE da árvore de decisão (exercício 4.2). Qual foi melhor?

```python
# 5.1
from sklearn.ensemble import RandomForestRegressor
modelo_rf = RandomForestRegressor(random_state=1)
modelo_rf.fit(X_train, y_train)
previsoes_rf = modelo_rf.predict(X_val)
mae_rf = mean_absolute_error(y_val, previsoes_rf)
mae_rf
```

```python
# 5.2
print(f"MAE árvore de decisão (melhor): {mae[best_max_leaf_nodes]:.2f}")
print(f"MAE random forest: {mae_rf:.2f}")
```
